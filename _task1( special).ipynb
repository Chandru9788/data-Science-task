```python
x = np.array([-5, 9, 20, 25, -3, 5, 16, 10])

x[(x >= -5) & (x <= 15)] *= -1

print(x)
```

```python
def array_info(arr):
    return (arr.shape, arr.ndim)

arr = np.array([0, 1, 2, 3, 4, 5])
print(array_info(arr))
```

```python
def first_last(arr):
    return (arr[0], arr[-1])

arr = np.array([0, 1, 2, 3, 4, 5])
print(first_last(arr))
```

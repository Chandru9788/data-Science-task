```python
import pandas as pd
pd.read_csv("netflix_titles.csv")
```

```python
net=pd.read_csv("netflix_titles.csv")
```

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
#1)Display the total number of Movies and TV Shows using a bar chart.
content_count = net['type'].value_counts()

plt.bar(content_count.index, content_count.values)

plt.title("Total Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Total Count")

plt.show()
```

```python
#2)Create a pie chart showing content percentages.
content_count = net['type'].value_counts()

plt.pie(content_count.values,
        labels=content_count.index,
        autopct='%1.1f%%',)

plt.title("Movies and TV Shows Percentage")

plt.show()
```

```python
#3)Create a histogram for movie durations.

movie = net[net['type'] == 'Movie']

duration = movie['duration'].str.replace(' min', '')

duration = pd.to_numeric(duration, errors='coerce').dropna()

plt.hist(duration, bins=20, edgecolor='black')

plt.title("Movie Duration Distribution")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Frequency")

plt.show()
```

```python
#4)Display the top 15 countries using a horizontal bar chart.
country_count = net['country'].dropna().str.split(',').explode()

top_countries = country_count.str.strip().value_counts().head(15)

plt.barh(top_countries.index, top_countries.values)

plt.title("Top 15 Countries")
plt.xlabel("Number of Content")
plt.ylabel("Country")

plt.gca().invert_yaxis()

plt.show()
```

```python
#5)Create a line chart showing content released per year.

release_year = net['release_year'].value_counts().sort_index()

plt.plot(release_year.index, release_year.values, marker='o')

plt.title("Content Released Per Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Content")

plt.show()
```

```python
#6)Compare Movies and TV Shows using two subplots.

movie = net[net['type'] == 'Movie']

tv = net[net['type'] == 'TV Show']

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)

plt.hist(movie['release_year'], bins=20, edgecolor= 'orange')

plt.title("Movies Released")
plt.xlabel("Year")
plt.ylabel("Count")


plt.subplot(1, 2, 2)

plt.hist(tv['release_year'], bins=20, edgecolor= 'red')

plt.title("TV Shows Released")
plt.xlabel("Year")
plt.ylabel("Count")

plt.tight_layout()

plt.show()
```

```python
#7)Find the top 10 most common content ratings and visualize them.

rating_count = net['rating'].value_counts().head(10)

plt.bar(rating_count.index, rating_count.values)

plt.title("Top 10 Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
#8)Create a box plot to identify movie duration outliers.

movie = net[net['type'] == 'Movie']

duration = movie['duration'].str.replace(' min', '')

duration = pd.to_numeric(duration, errors='coerce').dropna()

plt.boxplot(duration)

plt.title("Movie Duration Outliers")
plt.ylabel("Duration (Minutes)")

plt.show()
```

```python
#9)Compare the number of Movies and TV Shows released each year using two lines.

movie = net[net['type'] == 'Movie']

tv = net[net['type'] == 'TV Show']

movie_year = movie['release_year'].value_counts().sort_index()

tv_year = tv['release_year'].value_counts().sort_index()

plt.plot(movie_year.index, movie_year.values,
         label='Movies', marker='o')

plt.plot(tv_year.index, tv_year.values,
         label='TV Shows', marker='o')

plt.title("Movies vs TV Shows Released Each Year")

plt.xlabel("Release Year")
plt.ylabel("Number of Releases")

plt.legend()

plt.show()
```
